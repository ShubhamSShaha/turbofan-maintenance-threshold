# When should you service the engine?

An airline services every jet engine after **150 flights**, whether it is healthy or not.

- Some engines **break before** flight 150. That is very expensive.
- Most engines were **still fine** at 150. Good flying time was thrown away.

Each engine has a sensor that measures the temperature of the gas leaving the turbine. **A worn engine runs hotter**, so that temperature slowly climbs over the engine's life. Airlines watch exactly this kind of reading in real life.

**Your job: pick one temperature.** When an engine reaches your temperature, it goes for servicing. Then see how your answer changes when the cost of a breakdown changes.

| What happens | Cost |
|---|---|
| Serviced before it broke | **\$1M** |
| It broke in service | **\$5M** |
| Each good flight wasted by servicing too early | **\$0.02M** |

Pick too low and you waste good engines. Pick too high and engines break. Whoever costs the airline the least wins.

*Costs are illustrative. The engines are simulated (NASA turbofan data).*

```python
#@title Step 1: Run this cell first (about 5 seconds). Press the play button on the left.
_DATA = ""

import base64, gzip, io
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams.update({"figure.figsize": (9, 4.2), "axes.grid": True, "grid.alpha": 0.3,
                     "axes.spines.top": False, "axes.spines.right": False})

# The three costs (illustrative, in $ million)
COST_SERVICE_ON_TIME = 1.0    # engine serviced before it broke
COST_BREAKDOWN       = 5.0    # engine broke in service
COST_PER_WASTED_FLIGHT = 0.02 # each flight the engine could still have flown

engines = pd.read_csv(io.BytesIO(gzip.decompress(base64.b64decode(_DATA))))
# one odd reading should not fool us, so we use the average of the last 10 flights
engines["reading"] = engines.groupby("engine").temperature.transform(lambda s: s.rolling(10, min_periods=10).mean())
engines["raw"] = engines.temperature
_life = engines.groupby("engine").flight.max()


def _run(trigger=None, fixed=None, breakdown=COST_BREAKDOWN, col="reading"):
    rows = []
    for e, g in engines.groupby("engine"):
        life = len(g)
        if fixed is not None:
            at = fixed
        else:
            hit = np.where(g[col].to_numpy() >= trigger)[0]
            at = g.flight.to_numpy()[hit[0]] if len(hit) else None
        broke = at is None or at >= life
        wasted = 0 if broke else life - at
        rows.append((broke, wasted, breakdown if broke else COST_SERVICE_ON_TIME + COST_PER_WASTED_FLIGHT * wasted))
    r = pd.DataFrame(rows, columns=["broke", "wasted", "cost"])
    ok = r[~r.broke]
    return {"broke": int(r.broke.sum()), "on_time": len(ok),
            "wasted": float(ok.wasted.mean()) if len(ok) else 0.0, "cost": float(r.cost.sum())}


_TODAY = _run(fixed=150)


def show_engines(trigger=None, how_many=10, col="reading"):
    """Plot the turbine outlet temperature for a few engines. Every line is one engine; it breaks at the right edge."""
    fig, ax = plt.subplots()
    for e in list(engines.engine.unique())[:how_many]:
        g = engines[engines.engine == e]
        ax.plot(g.flight - g.flight.max(), g[col], lw=1.3 if col == "reading" else 0.8, alpha=0.85)
    if trigger is not None:
        ax.axhline(trigger, color="black", ls="--", lw=1.5)
        ax.text(ax.get_xlim()[0], trigger, f" your trigger: {trigger} deg C", va="bottom", fontsize=10)
    ax.set_xlabel("Flights left before the engine breaks  (0 = it breaks)")
    ax.set_ylabel("Turbine outlet temperature (deg C)")
    ax.set_title(f"{how_many} engines: the engine runs hotter as it wears out")
    plt.show()


def try_trigger(trigger, name="Your name", breakdown_cost=COST_BREAKDOWN, raw=False):
    """Service every engine as soon as its temperature reaches your trigger. What does it cost?"""
    col = "raw" if raw else "reading"
    show_engines(trigger, col=col)
    r = _run(trigger=trigger, breakdown=breakdown_cost, col=col)
    today = _run(fixed=150, breakdown=breakdown_cost)
    diff = today["cost"] - r["cost"]
    print(f"Your rule: service the engine when the temperature reaches {trigger} deg C"
          + ("  (raw reading, no averaging)" if raw else ""))
    print(f"A breakdown costs ${breakdown_cost:g}M\n")
    print(f"  Out of 100 engines:   {r['on_time']} serviced in time,  {r['broke']} broke")
    print(f"  Good flights wasted:  {r['wasted']:.0f} per serviced engine, on average")
    print(f"  Total cost:           ${r['cost']:.0f}M")
    print(f"  Today's schedule:     ${today['cost']:.0f}M  ({today['broke']} broke)")
    print(f"\n  You {'SAVE' if diff >= 0 else 'LOSE'} ${abs(diff):.0f}M compared with today.\n")
    print(f"Paste in chat  ->  {name}: breakdown ${breakdown_cost:g}M, trigger {trigger}, cost ${r['cost']:.0f}M, {r['broke']} broke")


def reveal(breakdown_costs=(1.5, 5, 20)):
    """The cost of every possible trigger temperature, for several breakdown costs."""
    ts = np.round(np.arange(506, 523.01, 0.25), 2)
    colors = ["#2a9d8f", "#3b6ea5", "#b0413e"]
    fig, ax = plt.subplots(figsize=(9, 4.8))
    for b, c in zip(breakdown_costs, colors):
        costs = [_run(trigger=t, breakdown=b)["cost"] for t in ts]
        i = int(np.argmin(costs))
        nb = _run(trigger=ts[i], breakdown=b)["broke"]
        ax.plot(ts, costs, color=c, lw=2.2,
                label=f"breakdown costs \\${b:g}M: cheapest at {ts[i]:g} deg C, {nb} of 100 break (\\${costs[i]:.0f}M)")
        ax.scatter([ts[i]], [costs[i]], color=c, edgecolor="black", zorder=3, s=55)
    ax.set_ylim(0, 520)
    ax.text(506.2, 20, "too early: wasting good engines", fontsize=9, va="bottom")
    ax.text(522.8, 20, "too late: engines break", fontsize=9, va="bottom", ha="right")
    ax.set_xlabel("Trigger temperature (deg C)")
    ax.set_ylabel("Total cost for 100 engines ($M)")
    ax.set_title("Same engines, same data. The dearer a breakdown, the earlier you service.")
    ax.legend(fontsize=9, loc="upper left", frameon=False)
    plt.show()

print(f"Ready. Loaded {engines.engine.nunique()} engines, each flown until it broke.")
```
